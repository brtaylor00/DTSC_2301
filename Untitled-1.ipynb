```python
variables = ",".join([
    "NAME",            # County name
    "B01001_001E",     # Total population
    "B19013_001E",     # Median household income (past 12 months)
    "B25077_001E",     # Median home value
    "B15003_022E"      # Bachelor's degree count
])

url = f"https://api.census.gov/data/2024/acs/acs1?get=NAME,B01001_001E&for=place:*&key=19ff7b05044205f420fd7590ff22343e29660e12"

response = requests.get(url)

data = response.json()

df = pd.DataFrame(data[1:], columns=data[0])

df.head()
```

```python
df.to_csv("census_data.csv", index=False)
```

```python
# Do we have nulls?
df = df.dropna()
df.isnull().sum()
```

```python
df.dtypes
df['state'] = df['state'].astype(int)
df['B01001_001E'] = df['B01001_001E'].astype(int)
df['place'] = df['place'].astype(int)
```

```python
numeric_cols = [
    "B01001_001E",
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce") # coerce -> If something cannot be converted to a number, don’t crash. Instead, turn it into NaN.

df = df.rename(columns={
    "B01001_001E": "Total_Population"
})
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x="state", y="Total_Population")
plt.title("Total Population by State")
plt.xlabel("State (in alphabetical order - Alabama to Wyoming)")
plt.ylabel("Total Population")
plt.show()
```

```python
top_population_cities = df.sort_values("Total_Population", ascending=False).head(10)

plt.figure(figsize=(8,6))
sns.barplot(data=top_population_cities, x="Total_Population", y="NAME")
plt.title("Top 10 Cities by Total Population")
plt.xlabel("Total Population (Millions)")
plt.ylabel("City")
plt.tight_layout()
plt.show()
```
